```python
#Task 1 : (2) List Operations
'''
o Create a list numbers with values: 10, 20, 30, 40, 50
o Create a second list more_numbers with values: 60, 70, 80
o Combine both lists into a new list called all_numbers
o Make a copy of all_numbers called numbers_copy
o Reverse the order of numbers_copy
o Print both all_numbers and numbers_copy
'''

# Create a list numbers with values: 10, 20, 30, 40, 50
numbers = [10, 20, 30, 40, 50]
print(numbers)
# Create a second list more_numbers with values: 60, 70, 80
more_numbers = [60, 70, 80]
print(more_numbers)

# Combine both lists into a new list called all_numbers
all_numbers = numbers + more_numbers
print("All Numbers:", all_numbers)

# Make a copy of all_numbers called numbers_copy
numbers_copy = all_numbers.copy()
print("Numbers Copy:", numbers_copy)

# Reverse the order of numbers_copy
numbers_copy.reverse()
print("Reversed Copy:", numbers_copy)



```

```python
#Task 1 : (3) Sorting and Counting

'''
o Create a list scores with these values: 85, 92, 78, 65, 92, 85, 74
o Sort the list in ascending order
o Sort the list in descending order
o Count how many times 92 appears in the list
o Find and print the highest and lowest scores
o Print the sorted list
'''

# Create a list scores with values
scores = [85, 92, 78, 65, 92, 85, 74]

# Sort the list in ascending order
scores.sort()
print("Ascending Order:", scores)

# Sort the list in descending order
scores.sort(reverse=True)
print("Descending Order:", scores)

# Count how many times 92 appears in the list
count_92 = scores.count(92)
print("92 appears:", count_92, "times")

# Find and print the highest and lowest scores
highest = max(scores)
lowest = min(scores)
print("Highest Score:", highest)
print("Lowest Score:", lowest)

# Print the sorted list (descending in this case)
print("Final Sorted List:", scores)

```

```python
#Task 2 : (3) Dictionary Operations
'''
o Create a second dictionary student_address with these key-value pairs:
 "street": "123 College Ave"
 "city": "New York"
 "zip": "10001"
o Combine both dictionaries into a new dictionary called student_info
o Remove the "zip" key from student_info
o Check if "phone" exists in the dictionary
o Print the final dictionary'''

# First dictionary
student = {
    "name": "John",
    "age": 21,
    "grade": "A+",
    "courses": ["Math", "Science", "History", "Computer Science"],
    "email": "john@example.com"
}

# Second dictionary
student_address = {
    "street": "123 College Ave",
    "city": "New York",
    "zip": "10001"
}
print(student, student_address)
# Combine both dictionaries into a new dictionary
student_info = {**student, **student_address}

# Remove the "zip" key
student_info.pop("zip")

# Check if "phone" exists in the dictionary
has_phone = "phone" in student_info
print("Does 'phone' exist in dictionary?", has_phone)

# Print the final dictionary
print("Final Student Info Dictionary:", student_info)
```

```python
#Task 2 : (4) Dictionary Methods
'''
o Create a dictionary word_count with these key-value pairs:
 "hello": 5
 "world": 10
 "python": 15
o Get all keys and print them as a list
o Get all values and print them as a list
o Get all key-value pairs and print them
o Make a copy of the dictionary
o Clear the original dictionary
o Print both dictionaries to verify one is empty and one is a copy
'''

# create dictionary
word_count = {
    "hello": 5,
    "world": 10,
    "python": 15
}

# Get all keys and print them as a list
keys_list = list(word_count.keys())
print("Keys:", keys_list)

# Get all values and print them as a list
values_list = list(word_count.values())
print("Values:", values_list)

# Get all key-value pairs and print them
items_list = list(word_count.items())
print("Key-Value Pairs:", items_list)

# Make a copy of the dictionary
word_count_copy = word_count.copy()

# Clear the original dictionary
word_count.clear()

# Print both dictionaries
print("Original Dictionary (after clear):", word_count)
print("Copied Dictionary:", word_count_copy)
```

```python
#Shopping Discount Calculator

#Task 3
'''
o purchase_amount: The total cost of items in the shopping cart
o is_loyalty_member: Boolean (True/False) indicating if the customer is a
loyalty member
o day_of_week: A string with the current day (e.g., "Monday", "Tuesday", etc.)
o is_holiday_season: Boolean indicating if it's currently a holiday shopping
season
2. Apply discount rules using if statements:
o If purchase amount is greater than $100, apply a 10% discount
o If purchase amount is greater than $50 (but less than or equal to $100),
apply a 5% discount
o If the customer is a loyalty member, apply an additional 5% discount
o If it's a weekend (Saturday or Sunday), apply an additional 2% discount
o If it's a holiday season, apply an additional 8% discount
3. Calculate and display:
o The original purchase amount
o Each discount that was applied and its amount
o The final price after all discounts
'''

# Initial variables
purchase_amount = 120.0
is_loyalty_member = True
day_of_week = "Sunday"
is_holiday_season = True

# discounts
original_amount = purchase_amount
total_discount = 0

# 1. Purchase amount discount
if purchase_amount > 100:
    discount = purchase_amount * 0.10
    print("10% discount for purchases over $100: -$", round(discount, 2))
    total_discount += discount
elif purchase_amount > 50:
    discount = purchase_amount * 0.05
    print("5% discount for purchases over $50: -$", round(discount, 2))
    total_discount += discount

# 2. Loyalty discount
if is_loyalty_member:
    loyalty_discount = purchase_amount * 0.05
    print("5% loyalty discount: -$", round(loyalty_discount, 2))
    total_discount += loyalty_discount

# 3. Weekend discount
if day_of_week == "Saturday" or day_of_week == "Sunday":
    weekend_discount = purchase_amount * 0.02
    print("2% weekend discount: -$", round(weekend_discount, 2))
    total_discount += weekend_discount

# 4. Holiday season discount
if is_holiday_season:
    holiday_discount = purchase_amount * 0.08
    print("8% holiday season discount: -$", round(holiday_discount, 2))
    total_discount += holiday_discount

# Final price
final_price = purchase_amount - total_discount

# Display summary
print("\nOriginal Purchase Amount: $", round(original_amount, 2))
print("Total Discounts Applied: -$", round(total_discount, 2))
print("Final Price to Pay: $", round(final_price, 2))

```

```python
#While Loop Assignment - Interactive Game

#Task 5
'''
Write a Python program that:
1. Sets a "secret number" (you can choose any number between 1 and 50)
2. Asks the player to guess the number
3. Tells the player if their guess is too high or too low
4. Keeps asking until the player guesses correctly
5. Keeps track of how many guesses it took
'''

import random

# 1. Set a secret number between 1 and 50
secret_number = random.randint(1, 50)

# 2. Initialize guess counter
guess_count = 0

print("Welcome to the Number Guessing Game!")
print("Guess the number between 1 and 50.")

# 3. Start the guessing loop
while True:
    try:
        guess = int(input("Enter your guess: "))
        guess_count += 1

        if guess < secret_number:
            print("Too low! Try again.")
        elif guess > secret_number:
            print("Too high! Try again.")
        else:
            print(f"Congratulations! You guessed it in {guess_count} tries.")
            break
    except ValueError:
        print("Please enter a valid number.")

```

```python
#Student Grade Management System

#Task 6

'''
1. Create a list to store information about students
2. Each student should be represented as a dictionary with the following keys:
o name: The student's full name
o id: A unique student ID number
o grades: Another dictionary containing course names as keys and grades as
values
3. Your program should:
o Add at least 3 students to the list
o Add grades for at least 3 courses for each student
o Display all students with their grades
o Calculate and display the average grade for each student
o Find and display the student with the highest average grade
o List all students with an average grade above 80
o Calculate and display the average grade for a specific course
o Add a new student to the system
'''

# Step 1: Create a list to store students
students = []

# Step 2: Add at least 3 students with grades for at least 3 courses
students.append({
    "name": "Syed Babar Mehmood Zaidi",
    "id": 101,
    "grades": {
        "Math": 85,
        "Science": 90,
        "English": 78
    }
})

students.append({
    "name": "Bamsi",
    "id": 102,
    "grades": {
        "Math": 92,
        "Science": 88,
        "English": 81
    }
})

students.append({
    "name": "Zain Malik",
    "id": 103,
    "grades": {
        "Math": 76,
        "Science": 83,
        "English": 69
    }
})

# Step 3: Display all students with their grades
print("\nAll Student Records with Grades:")
for student in students:
    print(f"ID: {student['id']} | Name: {student['name']} | Grades: {student['grades']}")

# Step 4: Calculate and display the average grade for each student
print("\nAverage Grade for Each Student:")
for student in students:
    grades = student['grades'].values()
    avg = sum(grades) / len(grades)
    student['average'] = avg  # store average in student dictionary
    print(f"{student['name']} (ID: {student['id']}): {avg:.2f}")

# Step 5: Find and display the student with the highest average grade
top_student = max(students, key=lambda x: x['average'])
print(f"\nTop Student: {top_student['name']} with average grade {top_student['average']:.2f}")

# Step 6: List all students with average grade above 80
print("\nStudents with Average Grade Above 80:")
for student in students:
    if student['average'] > 80:
        print(f"{student['name']} (Avg: {student['average']:.2f})")

# Step 7: Calculate and display the average grade for a specific course
course_name = "Math"
total = 0
count = 0

for student in students:
    if course_name in student['grades']:
        total += student['grades'][course_name]
        count += 1

if count > 0:
    course_avg = total / count
    print(f"\nAverage grade in {course_name}: {course_avg:.2f}")
else:
    print(f"\nNo grades found for course: {course_name}")

# Step 8: Add a new student to the system
new_student = {
    "name": "Areeba Noor",
    "id": 104,
    "grades": {
        "Math": 88,
        "Science": 91,
        "English": 85
    }
}
grades = new_student['grades'].values()
new_student['average'] = sum(grades) / len(grades)
students.append(new_student)
print(f"\nNew student added: {new_student['name']} (Avg: {new_student['average']:.2f})")

# Final list after adding new student
print("\nFinal List of Students:")
for student in students:
    print(f"ID: {student['id']} | Name: {student['name']} | Avg: {student['average']:.2f}")
```
